```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
```

```python
llm = ChatOpenAI(model = 'gpt-4.1-mini', temperature = 0, api_key=api_key)
response = llm.invoke('Explain about the importance of GITHUB in DataScience and AI for a fresher to get placed')
```

```python
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_template(
    """You are a {Profession}
    Explain {Topic} with respect to this profession in simple word. Make sure each points starts with '*' """
)
```

```python
prompt
```

```python
formatted_output = prompt.invoke({
    "Profession" : "Design Engineer in Aerospace" , "Topic": "Use of AI"}
                                )
```

```python
prompt_output = llm.invoke(formatted_output)
```

```python
print(prompt_output.content)
```

```python
# LangChain Expression Language (LCEL)

chain = prompt | llm
response = chain.invoke({
    "Profession" : "Marine Engineering" , "Topic": "Use of AI"})
print(response.content)        
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt2 = ChatPromptTemplate.from_template("Tomorrow is {Topic1} mock test tell the topics to cover")
parser =  StrOutputParser()
chain = prompt2 | llm | parser
topic = input(str())
response = chain.invoke({"Topic1" : topic})
print(response)
```
